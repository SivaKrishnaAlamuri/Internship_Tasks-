# Q1. Find the Shortest Word in a Sentence

```python
def find_shortest_word(sentence):
    words = sentence.split()
    if not words:
        return ""
    shortest_word = words[0]
    for word in words:
        if len(words) < len(shortest_word):
            shortest_word = word
    return shortest_word   

print(find_shortest_word("Python makes coding easy"))         
```

# Q5. Find the Word with the Maximum Number of Characters

```python
def find_max_char_word(sentence):
    words = sentence.split()
    if not words:
        return ""
    max_word = words[0]
    for word in words:
        if len(word) > len(max_word):
            max_word = word
    return max_word
print(find_max_char_word("Data engineering requires problem solving"))  
```

# Q6. Count Vowels in a String


```python
def count_vowels(s):
    vowels = "aeiouAEIOU"
    count = 0
    for char in s:
        if char in vowels:
            count += 1
    return count
print(count_vowels("Education"))  
```

# Q7. Count Consonants in a String

```python
def count_consonants(s):
    vowels = "aeiouAEIOU"
    count = 0
    for char in s:
        if char.isalpha() and char not in vowels:
            count += 1
    return count
print(count_consonants("Python 123!"))
```

# Q8. Find the Most Frequent Character


```python
def most_frequent_character(s):
    s = s.replace(" ", "")
    if not s:
        return ""
    freq = {}
    for char in s:
        freq[char] = freq.get(char, 0) + 1
    
    max_char = s[0]
    max_count = 0
    for char in s:
        if freq[char] > max_count:
            max_count = freq[char]
            max_char = char
    return max_char
print(most_frequent_character("banana"))
```

# Q9. Find the Least Frequent Character


```python
def least_frequent_character(s):
    s = s.replace(" ", "")
    if not s:
        return ""
    freq = {}
    for char in s:
        freq[char] = freq.get(char, 0) + 1
        
    min_char = s[0]
    min_count = float('inf')
    for char in s:
        if freq[char] < min_count:
            min_count = freq[char]
            min_char = char
    return min_char
print(least_frequent_character("success"))
```

# Q10. Find All Characters Appearing More Than Once


```python
def find_repeated_characters(s):
    s = s.replace(" ", "")
    freq = {}
    for char in s:
        freq[char] = freq.get(char, 0) + 1
        
    result = []
    for char in s:
        if freq[char] > 1 and char not in result:
            result.append(char)
    return result
print(find_repeated_characters("programming"))
```

# Q11. Create a Dictionary of Words and Their First Characters


```python
def words_first_chars(sentence):
    words = sentence.split()
    result = {}
    for word in words:
        if word:
            result[word] = word[0]
    return result
print(words_first_chars("Python Data Engineering"))
```

# Q12. Create a Dictionary of Words and Their Last Characters


```python
def words_last_chars(sentence):
    words = sentence.split()
    result = {}
    for word in words:
        if word:
            result[word] = word[-1]
    return result
print(words_last_chars("Python Data Engineering")) 
```

# Q13. Create a Dictionary of Words and Their Uppercase Versions


```python
def words_uppercase(sentence):
    words = sentence.split()
    result = {}
    for word in words:
        result[word] = word.upper()
    return result
print(words_uppercase("data science python"))
```

# Q14. Create a Dictionary of Numbers and Their Squares


```python
def number_squares(lst):
    result = {}
    for num in lst:
        result[num] = num ** 2
    return result
print(number_squares([2, 3, 4, 5]))
```

# Q15. Create a Dictionary of Numbers and Their Even/Odd Status


```python
def even_odd_status(lst):
    result = {}
    for num in lst:
        if num % 2 == 0:
            result[num] = "Even"
        else:
            result[num] = "Odd"
    return result
print(even_odd_status([1, 2, 3, 4, 5]))
```

# Q16. Find the Most Frequent Word in a Sentence


```python
def most_frequent_word(sentence):
    words = sentence.lower().split()
    if not words:
        return ""
    freq = {}
    for word in words:
        freq[word] = freq.get(word, 0) + 1
        
    max_word = words[0]
    max_count = 0
    for word in words:
        if freq[word] > max_count:
            max_count = freq[word]
            max_word = word
    return max_word
print(most_frequent_word("Python is easy and Python is powerful")) 
```

# Q17. Find All Words Appearing More Than Once


```python
def repeated_words(sentence):
    words = sentence.lower().split()
    freq = {}
    for word in words:
        freq[word] = freq.get(word, 0) + 1
        
    result = []
    for word in words:
        if freq[word] > 1 and word not in result:
            result.append(word)
    return result
print(repeated_words("data science uses data and science tools"))
```

# Q18. Find the First Word That Appears Only Once


```python
def first_unique_word(sentence):
    words = sentence.lower().split()
    freq = {}
    for word in words:
        freq[word] = freq.get(word, 0) + 1
        
    for word in words:
        if freq[word] == 1:
            return word
    return None
print(first_unique_word("apple banana apple mango banana orange"))
```

# Q19. Count Words Whose Length Is Greater Than 4


```python
def count_long_words(sentence):
    words = sentence.split()
    count = 0
    for word in words:
        if len(word) > 4:
            count += 1
    return count
print(count_long_words("Python makes problem solving interesting")) 
```

# Q20. Find the Longest Word That Appears Only Once

```python
def longest_unique_word(sentence):
    words = sentence.lower().split()
    freq = {}
    for word in words:
        freq[word] = freq.get(word, 0) + 1
        
    longest = ""
    for word in words:
        if freq[word] == 1:
            if len(word) > len(longest):
                longest = word
    return longest if longest else None
print(longest_unique_word("Python makes coding useful Python"))
```
