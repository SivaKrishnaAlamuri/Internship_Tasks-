## Q1. Find the Frequency of Each Word in a Sentence


```python
def word_frequency(sentence):
    frequency = {}
    for word in sentence.split():
        word = word.lower()
        frequency[word] = frequency.get(word, 0) + 1
    return frequency

print(word_frequency("python is easy and python is powerful"))
print(word_frequency("hello hello world"))
```

## Q2. Find the Frequency of Each Character in a Word


```python
def character_frequency(word):
    frequency = {}
    for char in word:
        frequency[char] = frequency.get(char, 0) + 1
    return frequency

print(character_frequency("banana"))
print(character_frequency("book"))
```

## Q5. Find the First Repeating Character in a String

```python
def first_repeating_character(string):
    seen = set()
    for char in string:
        if char in seen:
            return char
        seen.add(char)
    return None

print(first_repeating_character("programming"))
print(first_repeating_character("abcdefa"))
```

## Q6. Move All Zeros to the End of a List


```python
def move_zeros_to_end(nums):
    non_zeros = [number for number in nums if number != 0]
    zeros = [number for number in nums if number == 0]
    return non_zeros + zeros

print(move_zeros_to_end([0, 3, 0, 1, 2, 0, 4]))
print(move_zeros_to_end([1, 0, 2]))
```

## Q7. Find the Second Highest Distinct Number in a List


```python
def second_highest_distinct(nums):
    distinct_numbers = sorted(set(nums), reverse=True)
    if len(distinct_numbers) < 2:
        return None
    return distinct_numbers[1]

print(second_highest_distinct([10, 20, 40, 30, 40, 20]))
print(second_highest_distinct([5, 5, 4]))
```

## Q8. Remove Duplicates Without Changing the Original Order


```python
def remove_duplicates_preserve_order(nums):
    result = []
    seen = set()
    for number in nums:
        if number not in seen:
            result.append(number)
            seen.add(number)
    return result

print(remove_duplicates_preserve_order([4, 2, 4, 5, 2, 7, 5]))
print(remove_duplicates_preserve_order([1, 1, 1, 2]))
```

## Q9. Find the Missing Number from 1 to N

```python
def find_missing_number(nums, n):
    expected_sum = n * (n + 1) // 2
    return expected_sum - sum(nums)

print(find_missing_number([1, 2, 4, 5, 6], 6))
print(find_missing_number([1, 3, 4], 4))
```

## Q10. Find Two Numbers Whose Sum Equals a Target

```python
def two_sum_first_pair(nums, target):
    seen_numbers = []
    for current in nums:
        needed = target - current
        for earlier in seen_numbers:
            if earlier == needed:
                return [earlier, current]
        seen_numbers.append(current)
    return []

print(two_sum_first_pair([2, 7, 11, 15], 9))
print(two_sum_first_pair([1, 4, 6, 8], 10))
```

## Q11. Find the Number Closest to Zero


```python
def closest_to_zero(nums):
    if not nums:
        return None
    return min(nums, key=lambda number: (abs(number), -number))

print(closest_to_zero([-8, -3, 2, 5, -2]))
print(closest_to_zero([-4, -1, 1]))
```

## Q12. Find Common Elements Between Two Lists

```python
def common_elements(list1, list2):
    second_set = set(list2)
    result = []
    seen = set()
    for item in list1:
        if item in second_set and item not in seen:
            result.append(item)
            seen.add(item)
    return result

print(common_elements([1, 2, 3, 4, 5], [3, 4, 5, 6, 7]))
print(common_elements([2, 2, 3, 4], [4, 2]))
```

## Q13. Rotate a List to the Right by K Positions

```python
def rotate_right(nums, k):
    if not nums:
        return []
    k = k % len(nums)
    if k == 0:
        return nums[:]
    return nums[-k:] + nums[:-k]

print(rotate_right([1, 2, 3, 4, 5], 2))
print(rotate_right([1, 2, 3], 4))
```

## Q14. Find the Maximum Sum of Two Consecutive Elements


```python
def maximum_consecutive_pair_sum(nums):
    if len(nums) < 2:
        raise ValueError("The list must contain at least two elements.")
    maximum = nums[0] + nums[1]
    for index in range(1, len(nums) - 1):
        pair_sum = nums[index] + nums[index + 1]
        if pair_sum > maximum:
            maximum = pair_sum
    return maximum

print(maximum_consecutive_pair_sum([4, 7, 2, 9, 5]))
print(maximum_consecutive_pair_sum([-5, -2, -7]))
```

## Q15. Find All Pairs of Numbers Whose Sum Equals a Target

```python
def all_unique_pairs(nums, target):
    pairs = set()
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                pairs.add(tuple(sorted((nums[i], nums[j]))))
    return [list(pair) for pair in sorted(pairs)]

print(all_unique_pairs([2, 4, 3, 5, 7, 8, 1], 9))
print(all_unique_pairs([1, 2, 3, 4], 20))
```
