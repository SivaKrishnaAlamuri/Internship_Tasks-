```python
# Q35 Output:- 
# [[1, 99], [3, 4]]
# Explanation :- A shallow copy duplicates only the outer list; the inner lists are shared.

import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
b[0][1] = 99
print(a)
```

```python
# Q36 Output:- 
# [[1, 2], [3, 4]]
# Explanation :- A deep copy recursively copies nested lists, so modifying b's inner list does not affect a.

import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0][1] = 99
print(a)
```

```python
# Q37 Output:- 
# [[10], [20, 30]]
# Explanation :- list.copy() is shallow, so a and b share their inner lists.

import copy

a = [[10], [20]]
b = a.copy()
a[1].append(30)
print(b)
```

```python
# Q38 Output:- 
# [[10], [20]]
# Explanation :- deepcopy creates independent nested lists, so changes to a do not affect b.

import copy

a = [[10], [20]]
b = copy.deepcopy(a)
a[1].append(30)
print(b)
```

```python
# Q39 Output:- 
# false
# true
# Explanation :- The outer list is a new object, but its inner lists are shared by a shallow copy.

import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
print(a is b)
print(a[0] is b[0])
```
